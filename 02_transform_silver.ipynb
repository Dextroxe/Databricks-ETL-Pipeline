```python
from pyspark.sql.functions import col ,regexp_replace, year , current_date , concat_ws , when
```

```python
# Read Bronze
df = spark.table(bronze_table)
# df.display()
```

```python
#clean salary
df_cleaned = df.withColumn(
    "Salary",
    regexp_replace("Salary", "[$,]", "").cast("double")
)

# df_cleaned.display()
```

```python
df_partitioned = df_cleaned.withColumn(
    config["business_partition_column"],
    year(col("Hire_Date").cast("date"))
)

# df_partitioned.display()
```

```python
# processed date

df_processed = df_partitioned.withColumn(
    config["partition_columns"]["silver"],
    current_date()
)
```

```python
df_transformed = df_processed.withColumn(
    "full_name",
    concat_ws(" ", col("First_Name"), col("Last_Name"))
).withColumn(
    "salary_band",
    when(col("Salary") >= 80000, "High")
    .when(col("Salary") >= 70000, "Medium")
    .otherwise("Low")
)

# df_transformed.display()
```

```python
# write in table

df_transformed.write.mode("overwrite") \
    .partitionBy(config["business_partition_column"]) \
    .saveAsTable(silver_table)
```
