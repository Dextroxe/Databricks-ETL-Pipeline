```python
def load_config(path):
    try:
        with open(path, "r") as f:
            return json.load(f)
    except Exception as e:
        raise Exception(f"Error loading config file: {e}")

config = load_config(path)
```

```python
catalog = config["catalog"]
schema = config["schema"]
input_path = config["input_path"]

bronze_table = config["bronze_table"]
silver_table = config["silver_table"]
gold_agg_table = config["gold_agg_table"]
gold_detail_table = config["gold_detail_table"]

sensitive_columns = config["sensitive_columns"]
bronze_partition_col = config["partition_columns"]["bronze"]
silver_partition_col = config["partition_columns"]["silver"]


# Debug print (optional)
print("Config loaded successfully")
print("Input Path:", input_path)
print("Bronze Table:", f"{catalog}.{schema}.{bronze_table}")
```
